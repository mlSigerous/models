```python
from catboost import CatBoostClassifier
```

```python
data_cb = data.copy()

y = data_cb['Churn']
X = data_cb.drop(columns='Churn')

X_train, X_val, y_train, y_val = train_test_split(X, y,
                                                  shuffle=True,
                                                  test_size=0.2,
                                                  random_state=42,
                                                  stratify=y #этот параметр делает разбиение параметров на трейн и валидейшн с сохранением дисбаланса классов
                                                  #именно это в итоге дало мне +0.02 на ROC AUC даже со стандартными параметрами катбуста
                                                  )
```

Сначала обучим со стандартными параметрами

```python
model = CatBoostClassifier(random_seed=42)

model.fit(X_train, y_train, cat_features=cat_cols)

y_val_pred = model.predict_proba(X_val)[:, 1]
```

```python
roc_auc_score(y_val, y_val_pred)
```

Неплохо, но явно можно лучше. Подберем лучшие параметры

```python
#Это ОЧЕНЬ неоптимизированный код, параметры искались полтора часа. Дальше попытаюсь оптимизировать
model = CatBoostClassifier(
    cat_features=cat_cols,
    eval_metric='AUC',
    boosting_type='Ordered', #этот параметр лучше обрабатывает кат. признаки на небольших датасетах
    random_seed=42)

param_dist = {
    'depth': [4, 5, 6, 7, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'iterations': [1000, 2000, 3000],
    'l2_leaf_reg': [1, 3, 5, 10]}#параметр для регуляризации

search_result = model.randomized_search(
    param_distributions=param_dist,
    X=X_train,
    y=y_train,
    cv=3, #то есть мы можем сделать кросс-валидацию через встроенный в катбуст метод рандомсерча
    n_iter=15,
    search_by_train_test_split=False) #подбор гиперпараметров будет идти на кроссвалидации

y_val_pred = model.predict_proba(X_val)[:, 1]
```

```python
print(f'Score: {round(roc_auc_score(y_val, y_val_pred), 2)}\nBest params: {search_result["params"]}')
```
